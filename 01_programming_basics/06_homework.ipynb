```python
# 1. Print numbers from 100 to 1 using a loop.
for i in range(100,0,-1):
    print(i)
```

```python
# 2. Print all even numbers from 1 to 100.
for i in range(2,101,2):
    print(i)
```

```python
# 3. Reverse a list without using [::-1] or reverse(). 
arr = [10,20,30,40,50,60]
rev_arr = []
for i in range(len(arr)-1,-1,-1):
    rev_arr.append(arr[i])
print(rev_arr)
```

```python
# 4. Print the multiplication table of 7 using a loop. 
n = 7
for i in range(1,11):
    print(f'{n} X {i} = {n*i}')
```

```python
# 5. Find the largest element in a list using only a loop. 
arr = [10,20,30,40,50,22,58,399,22,33,18]
large = arr[0]
for i in arr:
    if i > large:
        large = i

print(int(large))
```
