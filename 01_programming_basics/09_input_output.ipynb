
#  Input → Process → Output in DSA

## Introduction

Every programming problem follows the same workflow:

```text
Input
   ↓
Process (Algorithm)
   ↓
Output
```

Before writing code, always understand these three parts.

---

# 1. What is Input?

**Input** is the data given to your program or function.

It can be:

* Integer
* Float
* String
* Array (List)
* Matrix
* Linked List
* Tree
* Graph

### Example

```python
def square(n):
    return n * n
```

Input

```text
5
```

Here,

```python
n = 5
```

---

# 2. What is Process?

The **Process** is the logic or algorithm used to solve the problem.

It converts the input into the required output.

Example

```python
def square(n):
    return n * n
```

Process

```text
5 × 5
```

---

# 3. What is Output?

The **Output** is the final result produced after processing the input.

Example

```python
print(square(5))
```

Output

```text
25
```

```text
Input

↓

Think

↓

Algorithm

↓

Write Code

↓

Output
```

# DSA Thinking Pattern

Whenever you solve a DSA problem, ask yourself these three questions.

## Question 1

### What is the Input?

Examples

```text
Integer

Array

String

Matrix

Tree

Graph
```

---

## Question 2

### What is the Expected Output?

Examples

```text
Largest Element

Smallest Element

True / False

Index

Count

Sorted Array

Maximum Sum
```

---

## Question 3

### How can I convert the Input into the Output?

This step is called the **Algorithm**.

---

# Real Example

Question:

Find the largest element in an array.

Input

```text
[10, 20, 80, 40]
```

Process

```text
Take the first element as the current largest.

Compare every remaining element.

Update the largest value whenever a bigger element is found.
```

Output

```text
80
```

---

# Why is Input → Process → Output Important?

Most beginners start writing code immediately.

Professional programmers first understand:

* What data is given?
* What result is expected?
* What steps are required?

Only then do they write code.

---

# Interview Strategy

Before writing a single line of code, answer these three questions:

### 1. What is the Input?

Example:

```text
Array
```

---

### 2. What is the Output?

Example:

```text
Second Largest Element
```

---

### 3. What is the Algorithm?

Example:

```text
Traverse the array.

Keep track of the largest and second largest values.

Return the second largest.
```

---

# Golden Rule

**Never start coding before understanding the Input, Process, and Output.**

This habit will make solving DSA problems much easier.
