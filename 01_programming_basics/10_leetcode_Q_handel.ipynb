# Step 1 — Never Start Coding Immediately

Suppose LeetCode gives you:

> Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

A beginner sees this:

```python
def twoSum(nums, target):
```

and immediately starts coding. 

Instead, first break the question down.

---

# Step 2 — Find the INPUT

Look for words like:

> **Given an array of integers `nums` and an integer `target`...**

That tells us our inputs are:

```text
nums   → Array/List
target → Integer
```

Example:

```text
nums = [2, 7, 11, 15]
target = 9
```

So mentally:

```text
INPUT
  ↓
[2, 7, 11, 15], 9
```

No algorithm yet.

---

# Step 3 — Find the OUTPUT

Now read:

> **return indices of the two numbers...**

Very important.

It does NOT say:

> return the numbers 

It says:

> return their **indices** 

For:

```text
nums = [2, 7, 11, 15]
```

Indices are:

```text
Index:  0  1   2   3
        ↓  ↓   ↓   ↓
Value: [2, 7, 11, 15]
```

Since:

```text
2 + 7 = 9
```

we return:

```text
[0, 1]
```

Not:

```text
[2, 7] 
```

This is exactly why carefully reading the **output requirement** matters.

---

# Step 4 — Find the CONDITION

Now read:

 **such that they add up to `target`**

That's our condition.

```text
number1 + number2 == target
```

For example:

```text
2 + 7 == 9
```

True. 

So now we know:

```text
INPUT
↓
nums = [2,7,11,15]
target = 9


CONDITION
↓
nums[i] + nums[j] == target


OUTPUT
↓
[i, j]
```

See? We haven't written any code yet, but we already understand the problem.

---

# Step 5 — Read the Example Properly

LeetCode usually provides something like:

```text
Input: nums = [2,7,11,15], target = 9

Output: [0,1]

Explanation:
Because nums[0] + nums[1] == 9,
we return [0,1].
```

Don't just look at the output.

Ask:

**Why did `[0,1]` come out?**

Because:

```text
nums[0] = 2
nums[1] = 7

2 + 7 = 9
```

This is called **dry-running the example**.

---

# Step 6 — Read Constraints

Now you'll see something like:

```text
2 <= nums.length <= 10⁴
-10⁹ <= nums[i] <= 10⁹
-10⁹ <= target <= 10⁹
```

Right now these numbers may look scary.

But constraints basically tell us:

> **How big can the input become?**

Why does that matter?

Because an algorithm that works for:

```text
10 elements
```

might become extremely slow for:

```text
10,000,000 elements
```

Constraints help us decide whether we need:

```text
O(n²)
O(n)
O(log n)
```

etc.

We'll learn this deeply with **Time Complexity**.

For now, just remember:

> **Constraints tell us how efficient our solution needs to be.**

---

# Step 7 — Look for Special Guarantees

This is VERY important.

Two Sum says something like:

> You may assume that each input would have exactly one solution.

That's valuable information.

It means we don't need to worry about:

```text
No solution 

Multiple solutions 
```

Exactly one valid pair exists.

Another line:

> You may not use the same element twice.

Meaning if:

```text
nums = [3, 2, 4]
target = 6
```

You cannot do:

```text
nums[0] + nums[0]

3 + 3 = 6 
```

because that's the **same element/index twice**.

But:

```text
nums = [3, 3]
```

is different.

```text
nums[0] + nums[1]

3 + 3 = 6 
```

Two different indices.

These small sentences often contain the most important information in a LeetCode question.

# Our LeetCode Question Reading Formula 

From today, whenever you open a LeetCode problem, don't code.

First write:

```text
1. INPUT:
   What data am I receiving?

2. OUTPUT:
   What exactly should I return?

3. CONDITION:
   What condition must be satisfied?

4. EXAMPLE:
   Why does the given input produce that output?

5. CONSTRAINTS:
   How large can the input become?

6. EDGE CASES:
   What unusual inputs could occur?

7. ALGORITHM:
   How can I convert the input into the required output?

8. COMPLEXITY:
   How much time and memory does my solution use?
```

For now, focus mainly on the **first 6**. Algorithm recognition will come gradually.
