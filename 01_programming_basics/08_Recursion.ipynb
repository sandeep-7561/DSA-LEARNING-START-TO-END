Chapter 1: What is Recursion?
Definition
Recursion is a technique where a function calls itself to solve a smaller version of the same problem.
In simple words:
Instead of using a loop (for or while), the function keeps calling itself until the work is finished.

Example:----------------------------
def hello():
    print("Hello")
    hello()

hello()
-----------------------------------
Output:
Hello
Hello
Hello
Hello
...
RecursionError
------------------------------------


Every Recursion Needs Two Things

1. Base Case 
The condition where recursion stops.

Example:----------
if n == 0:
    return
------------------

Without this,
Infinite recursion
↓
RecursionError

==================================
2. Recursive Call 

The function calling itself.
Example:---------------
countdown(n-1)
-----------------------

This moves the problem toward the base case.


First Correct Recursion: -----------------
def countdown(n):

    if n == 0:
        return

    print(n)

    countdown(n-1)

countdown(5)
-------------------------------------------

When Should We Use Recursion?
Use recursion when the problem can naturally be broken into smaller versions of itself.

Example:

Factorial(5)

↓

5 × Factorial(4)

↓

4 × Factorial(3)

↓

3 × Factorial(2)

↓

2 × Factorial(1)

Notice how the same function keeps solving a smaller problem.

Lets solv the problems of recurison: (Solve these without using loops)


```python
# 1. Print numbers from 1 to N. def print_numbers(n):. Solve these without using loops

number = 1
def print_numbers(n):
    global number 
    if n == 0:
        return
    print(number)
    number += 1
    print_numbers(n-1)


print_numbers(5)
```

```python
# 2. Print numbers from N to 1. def reverse_print(n):. Solve there without using loops

def reverse_print(n):
    if n == 0:
        return

    print(n)
    reverse_print(n-1)

reverse_print(5)
```

```python
# 3. Find the sum of the first N numbers. def sum_n(n): . Solve there without using loops
def sum_n(n):
    if n == 0:
        return 0
    else:
        return n + sum_n(n-1)

print(sum_n(5))
```
