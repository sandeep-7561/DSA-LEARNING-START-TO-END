```python
# 1. create a simple calculator its only add number without change function parameters add more numbers, for example: print(add(1,2,3)),print(add(1,2,3,4,5))

def calculator(*numbers):
    total = 0
    for i in numbers:
        total += i
    return total

print(calculator(1,2,3))
print(calculator(1,2,3,4,5))
```

```python
# 2. Find the Maximum Element. Write a functions: def find_max(arr): 
# Example:
'''
arr = [10, 50, 20, 90, 40]
print(find_max(arr))
output 90
'''
# Concepts: Function, Loop, Return, Comparison

def find_max(arr):
    valu_count = arr[0]
    for value in arr:
        if value > valu_count:
            valu_count = value

    return valu_count

arr = [10, 50, 20, 90, 40]
print(find_max(arr))


'''
Time Complexity
O(n)

Space Complexity
O(1)
'''

```

```python
# 3. count even numbers. Write a function: def count_even(arr) 
'''
Example:
arr = [1,2,3,4,5,6]

print(count_even(arr))

output 3
'''
# Concepts, Function, Loop, Modulus (%), Counter

def count_even(arr):
    total_even = 0
    for value in arr:
        if value%2 == 0:
            total_even += 1

    return total_even

arr = [1,2,3,4,5,6]
print(count_even(arr))


'''
Complexity

Time : O(n)
Space : O(1)
'''
```

```python
# 4. everse a List (Without reverse() or [::-1]). Write a function: def reverse_list(arr):
''' 
Example:
arr = [1,2,3,4]
print(reverse_list(arr))

Output [4,3,2,1]
'''
# Concepts: Loop, Indexing, Functions
def reverse_list(arr):
    rev_list = []
    for i in range(len(arr)-1,-1,-1):
        rev_list.append(arr[i])
    return rev_list

arr = [1,2,3,4]
print(reverse_list(arr))
```

```python
# 5. Check if a Number is Prime. Write a function: def is_prime(n):.
'''
Exaple:
print(is_prime(17))
print(is_prime(18))

output:
True
False
'''
# Concepts: Loop, Function, Conditions, Prime checking is one of the most common interview questions.

def is_prime(n):
    if n < 2:
        return False
    for i in range(2,n):
        if n%i == 0:
            return False
    return True
    
print(is_prime(17))
print(is_prime(12))
```

```python
# 6. Find the Second Largest Element. Write a function: def second_largest(arr):.
'''
Example:
arr = [10,40,30,90,70]
print(second_largest(arr))

Output:
70
'''
# Concepts: Functions, Variables, Comparison, Loop, This is a very common Amazon, Microsoft, and Infosys interview question.

def second_largest(arr):
    first_largest = arr[0]
    second_largest_num = arr[0]
    for i in arr:
        if i > first_largest:
            first_largest = i

    for i in arr:
        if i > second_largest_num and i < first_largest:
            second_largest_num = i
            
            
    return second_largest_num

arr = [10,40,30,90,70]
print(second_largest(arr))
```
