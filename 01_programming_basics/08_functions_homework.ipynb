```python
# 1. create a simple calculator its only add number without change function parameters add more numbers, for example: print(add(1,2,3)),print(add(1,2,3,4,5))

def calculator(*numbers):
    total = 0
    for i in numbers:
        total += i
    return total

print(calculator(1,2,3))
print(calculator(1,2,3,4,5))
```

```python
# Find the Maximum Element. Write a functions: def find_max(arr): 
# Example:
'''
arr = [10, 50, 20, 90, 40]
print(find_max(arr))
output 90
'''
# Concepts: Function, Loop, Return, Comparison

def find_max(arr):
    maxim_num = arr[0]
    for value in arr:
        if value > maxim_num:
            maxim_num = value

    return maxim_num

arr = [10,50,20,90,40]
print(find_max(arr))
```
