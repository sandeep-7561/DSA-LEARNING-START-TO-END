# Chapter 1: What is Time Complexity?

## Simple Definition

> **Time Complexity measures how the running time of an algorithm grows as the input size increases.**

Simple English:

It **does NOT measure seconds or milliseconds**.

It measures **how the number of operations increases when the input becomes larger.**

---

> Count all students in a classroom.

```text
1000 seconds
```

See the pattern?

```text
Students ↑

↓

Work ↑
```

This increase in work is what Time Complexity studies.

# Very Important

```text
"This code takes 2 seconds."
```

Wrong.

Because:

* Your PC
* My server
* Google's computer

all have different speeds.

Instead we say:

```text
If input becomes bigger,
how much extra work does the algorithm do?
```

# One Important Rule

Ignore:

* Constants
* Smaller terms

Example

```text
O(2n)
```

becomes

```text
O(n)
```

Because when `n` becomes huge, the constant `2` doesn't change the growth pattern.

Similarly

```text
O(n + 5)
```

becomes

```text
O(n)
```

And

```text
O(n² + n)
```

becomes

```text
O(n²)
```

We always keep the **dominant term**.

# Example 3

Reverse List

```python
def reverse(arr):

    ans = []

    for i in range(len(arr)-1,-1,-1):
        ans.append(arr[i])

    return ans
```
