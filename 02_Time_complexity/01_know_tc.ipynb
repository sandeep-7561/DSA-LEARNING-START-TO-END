# Chapter 1: What is Time Complexity?

## Simple Definition

> **Time Complexity measures how the running time of an algorithm grows as the input size increases.**

Simple English:

It **does NOT measure seconds or milliseconds**.

It measures **how the number of operations increases when the input becomes larger.**

---

# WhatsApp Meaning 

Suppose I ask you:

> Count all students in a classroom.

### Classroom A

```text
10 students
```

It may take

```text
10 seconds
```

---

### Classroom B

```text
100 students
```

Now it may take

```text
100 seconds
```

---

### Classroom C

```text
1000 students
```

Now it may take

```text
1000 seconds
```

See the pattern?

```text
Students ↑

↓

Work ↑
```

This increase in work is what Time Complexity studies.

---

# Very Important

Time Complexity is **NOT**

```text
"This code takes 2 seconds."
```

Wrong.

Because:

* Your PC
* My server
* Google's computer

all have different speeds.

Instead we say:

```text
If input becomes bigger,
how much extra work does the algorithm do?
```

# O(log n)

This one confuses everyone.

Imagine a phone book.

Need to find

```text
Rahul
```

Will you start from page 1?

No.

You open the middle.

If Rahul is after the middle,

throw away half the book.

Again open the middle.

Again remove half.

Again remove half.

This is

```text
Binary Search
```

Every step removes half.

Example

```text
1000

↓

500

↓

250

↓

125

↓

62

↓

31

↓

15

↓

7

↓

3

↓

1
```

Very fast.

Time Complexity

```text
O(log n)
```

---

# O(n log n)

Algorithms like

* Merge Sort
* Heap Sort

take

```text
O(n log n)
```

Much faster than

```text
O(n²)
```

for large inputs.

# One Important Rule

Ignore:

* Constants
* Smaller terms

Example

```text
O(2n)
```

becomes

```text
O(n)
```

Because when `n` becomes huge, the constant `2` doesn't change the growth pattern.

Similarly

```text
O(n + 5)
```

becomes

```text
O(n)
```

And

```text
O(n² + n)
```

becomes

```text
O(n²)
```

We always keep the **dominant term**.

# Common Time Complexities

| Code               | Complexity |
| ------------------ | ---------- |
| `print(x)`         | O(1)       |
| `arr[i]`           | O(1)       |
| One loop           | O(n)       |
| Two nested loops   | O(n²)      |
| Three nested loops | O(n³)      |
| Binary Search      | O(log n)   |
| Merge Sort         | O(n log n) |

# Rule 2: Nested Loop Means Multiplication

Example

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

Suppose

```text
n = 3
```

Outer Loop

```text
0
1
2
```

For every outer iteration,

Inner Loop runs

```text
0
1
2
```

Visualization

```text
i=0

j=0
j=1
j=2

------------

i=1

j=0
j=1
j=2

------------

i=2

j=0
j=1
j=2
```

Total

```text
3 × 3

=

9
```

Pattern

```text
n × n

=

n²
```

Complexity

```text
O(n²)
```

# Rule 3: Separate Loops = Addition

Example

```python
for i in range(n):
    print(i)

for j in range(n):
    print(j)
```

First loop

```text
n
```

Second loop

```text
n
```

Total

```text
n + n

=

2n
```

Big O ignores constants

```text
O(2n)

↓

O(n)
```

## Common Beginner Mistake

Many beginners think

```python
for i in range(n):
    print(i)

for j in range(n):
    print(j)
```

is

```text
O(n²)
```

Wrong.

Why?

Because one loop finishes **before** the second starts.

There is **no nesting**.

---

# Rule 4: Ignore Constants

Example

```python
for i in range(100):
    print(i)
```

How many times?

```text
100
```

Will it ever become 1000?

No.

No matter how large `n` is, this loop always runs exactly 100 times.

Complexity

```text
O(1)
```

Constant Time.

---

Another example

```python
for i in range(2*n):
    print(i)
```

Operations

```text
2n
```

Big O

```text
O(n)
```

The constant `2` is ignored.

---

# Rule 5: Ignore Smaller Terms

Example

```text
O(n² + n)
```

Which grows faster?

```text
n²
```

or

```text
n
```

Suppose

```text
n = 10
```

```text
n² = 100

n = 10
```

Suppose

```text
n = 1000
```

```text
n² = 1,000,000

n = 1000
```

Clearly,

```text
n²
```

dominates.

Therefore

```text
O(n² + n)

↓

O(n²)
```
