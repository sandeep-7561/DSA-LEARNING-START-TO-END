# Chapter 1: What is Time Complexity?

## Simple Definition

> **Time Complexity measures how the running time of an algorithm grows as the input size increases.**

Simple English:

It **does NOT measure seconds or milliseconds**.

It measures **how the number of operations increases when the input becomes larger.**

---

# WhatsApp Meaning

> Count all students in a classroom.

```text
1000 students
```

Now it may take

```text
1000 seconds
```

See the pattern?

```text
Students ↑

↓

Work ↑
```

This increase in work is what Time Complexity studies.

---

# Very Important

Time Complexity is **NOT**

```text
"This code takes 2 seconds."
```

Wrong.

Because:

* Your PC
* My server
* Google's computer

all have different speeds.

Instead we say:

```text
If input becomes bigger,
how much extra work does the algorithm do?
```

Example

```python
arr = [10,20,30]

print(arr[1])
```

Python directly goes to index 1.

```text
3 elements
```

# O(n²)

Nested loop

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

Let's dry run.

Suppose

```text
n = 3
```

Outer loop

```text
0
1
2
```

Inner loop

```text
0
1
2
```

Total

```text
3 × 3 = 9
```

Now

```text
n = 100
```

Operations

```text
100 × 100

=

10000
```

Pattern

```text
n × n

=

n²
```

Time Complexity

```text
O(n²)
```

---

# Visual

```text
n = 4

Outer

0
1
2
3

For EVERY outer loop

↓

Inner

0
1
2
3
```

So

```text
4 × 4

=

16 operations
```

# O(log n)

This one confuses everyone.

Imagine a phone book.

Need to find

```text
Rahul
```

Will you start from page 1?

No.

You open the middle.

If Rahul is after the middle,

throw away half the book.

Again open the middle.

Again remove half.

Again remove half.

This is

```text
Binary Search
```

Every step removes half.

Example

```text
1000

↓

500

↓

250

↓

125

↓

62

↓

31

↓

15

↓

7

↓

3

↓

1
```

Very fast.

Time Complexity

```text
O(log n)
```

---

# O(n log n)

Algorithms like

* Merge Sort
* Heap Sort

take

```text
O(n log n)
```

Much faster than

```text
O(n²)
```

for large inputs.

---

# One Important Rule

Ignore:

* Constants
* Smaller terms

Example

```text
O(2n)
```

becomes

```text
O(n)
```

Because when `n` becomes huge, the constant `2` doesn't change the growth pattern.

Similarly

```text
O(n + 5)
```

becomes

```text
O(n)
```

And

```text
O(n² + n)
```

becomes

```text
O(n²)
```

We always keep the **dominant term**.

---

# Common Time Complexities

| Code               | Complexity |
| ------------------ | ---------- |
| `print(x)`         | O(1)       |
| `arr[i]`           | O(1)       |
| One loop           | O(n)       |
| Two nested loops   | O(n²)      |
| Three nested loops | O(n³)      |
| Binary Search      | O(log n)   |
| Merge Sort         | O(n log n) |
