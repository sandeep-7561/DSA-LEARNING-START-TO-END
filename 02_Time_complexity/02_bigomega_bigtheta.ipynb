# Big O, Big Omega (Ω), and Big Theta (Θ)

## Introduction

When analyzing an algorithm, we want to know **how its performance changes as the input size increases**.

There are three common asymptotic notations used to describe an algorithm's performance:

* **Big O (O)** → Worst Case
* **Big Omega (Ω)** → Best Case
* **Big Theta (Θ)** → Average Case (or Tight Bound)

---

# 1. Big O Notation (O)

## Definition

> **Big O notation describes the maximum amount of time an algorithm may take to complete.**

It represents the **Worst Case Time Complexity**.

### Question it answers

> **"What is the maximum amount of work this algorithm may perform?"**

---

## Example: Linear Search

```python
def linear_search(arr, target):

    for num in arr:
        if num == target:
            return True

    return False
```

Suppose

```text
arr = [10, 20, 30, 40, 50]
target = 50
```

The algorithm checks:

```text
10

↓

20

↓

30

↓

40

↓

50
```

Every element must be visited.

Time Complexity

```text
O(n)
```

---

# 2. Big Omega Notation (Ω)

## Definition

> **Big Omega notation describes the minimum amount of time an algorithm needs to complete.**

It represents the **Best Case Time Complexity**.

### Question it answers

> **"What is the least amount of work this algorithm may perform?"**

---

## Example: Linear Search

```text
arr = [10, 20, 30, 40, 50]

target = 10
```

The first element is already the target.

The algorithm performs only one comparison.

Time Complexity

```text
Ω(1)
```

---

# 3. Big Theta Notation (Θ)

## Definition

> **Big Theta notation describes the average or exact growth rate of an algorithm.**

It represents the **Average Case Time Complexity** (or Tight Bound).

### Question it answers

> **"How does the algorithm usually perform?"**

---

## Example: Linear Search

Sometimes the target is:

* At the beginning
* In the middle
* At the end

On average, approximately **n/2** elements are checked.

Ignoring constants,

```text
n / 2

↓

n
```

Time Complexity

```text
Θ(n)
```

---

# Visual Comparison

Consider the following array:

```text
[10, 20, 30, 40, 50]
```

### Best Case

```text
Target = 10

↓

Found immediately

↓

Ω(1)
```

---

### Average Case

```text
Target = 30

↓

Approximately half of the array is checked

↓

Θ(n)
```

---

### Worst Case

```text
Target = 50

↓

Every element is checked

↓

O(n)
```

---

# Another Example: Binary Search

Binary Search repeatedly divides the search space into half.

### Best Case

The middle element is the target.

```text
Ω(1)
```

---

### Average Case

```text
Θ(log n)
```

---

### Worst Case

```text
O(log n)
```

---

# Comparison Table

| Notation          | Meaning                    | Represents           |
| ----------------- | -------------------------- | -------------------- |
| **Big O (O)**     | Worst Case                 | Maximum running time |
| **Big Omega (Ω)** | Best Case                  | Minimum running time |
| **Big Theta (Θ)** | Average Case (Tight Bound) | Typical running time |

---

# Real Interview Perspective

In coding interviews, interviewers almost always ask:

> **"What is the time complexity of your solution?"**

Unless specified otherwise, they are asking for the **Worst Case Time Complexity**, which is represented using **Big O notation**.

Example:

```text
Linear Search

Worst Case

↓

O(n)
```

---

# Why Do We Mostly Use Big O?

Big O is the most commonly used notation because it guarantees the algorithm's performance even in the worst possible situation.

When designing software, developers usually prepare for the **worst-case scenario**, making Big O the standard notation in interviews and competitive programming.

---

# Key Differences

| Feature              | Big O        | Big Omega    | Big Theta                  |
| -------------------- | ------------ | ------------ | -------------------------- |
| Represents           | Worst Case   | Best Case    | Average Case / Tight Bound |
| Focus                | Maximum work | Minimum work | Typical work               |
| Common in Interviews |   Yes        | Rarely       | Rarely                     |
| Used in LeetCode     |   Yes        | Sometimes    | Sometimes                  |

---

# Important Notes

* Big O describes the **upper bound** of an algorithm.
* Big Omega describes the **lower bound** of an algorithm.
* Big Theta describes the **tight (exact) bound** of an algorithm.
* In almost all coding interviews and LeetCode discussions, **Big O** is the standard notation used to describe Time Complexity.

---

# Summary

```text
Big Omega (Ω)
      ↑
   Best Case

Big Theta (Θ)
      ↑
 Average Case

Big O (O)
      ↑
 Worst Case
```
