```python
# 2. O(log n)

arr = [10,20,30,40,50,60,70,80]
target = 70
def binary_serch(arr,target):
    low = 0
    high = len(arr) -1
    while low <= high:
        mid = (low+high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid -1
    return -1

print(f"index : {binary_serch(arr,target)}")
print(f"value : {arr[binary_serch(arr,target)]}")


    
```
