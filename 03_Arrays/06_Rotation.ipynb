# Array Rotation in Data Structures & Algorithms

## Introduction

Array rotation is one of the most common interview topics in Data Structures and Algorithms.

Unlike insertion or deletion, rotation does not remove or add elements. Instead, it shifts the position of every element in the array.

Array rotation is frequently used in:

* Circular Arrays
* Queue Implementations
* Ring Buffers
* Scheduling Algorithms
* LeetCode Interview Problems

---

# What is Array Rotation?

## Definition

> **Array rotation is the process of shifting all elements of an array either to the left or to the right by a specified number of positions (k).**

The relative order of elements changes, but no element is lost.

---

# Types of Array Rotation

There are two types of rotation:

* Left Rotation
* Right Rotation

---

# Left Rotation

In a left rotation, every element moves one position to the left.

The first element moves to the last position.

### Example

Original Array

```python
arr = [1, 2, 3, 4, 5]
```

Rotate Left by 1

```python
[2, 3, 4, 5, 1]
```

---

# Right Rotation

In a right rotation, every element moves one position to the right.

The last element moves to the first position.

### Example

Original Array

```python
arr = [1, 2, 3, 4, 5]
```

Rotate Right by 1

```python
[5, 1, 2, 3, 4]
```

---

# Rotation by K Positions

Instead of rotating only once, an array can be rotated multiple times.

### Example

```python
arr = [1, 2, 3, 4, 5]

k = 2
```

Rotate Right

```python
[4, 5, 1, 2, 3]
```

---

# Brute Force Approach

## Idea

Rotate the array **one position at a time**, and repeat the process **k times**.

---

## Algorithm

1. Repeat the rotation **k** times.
2. Remove the last element.
3. Insert it at the beginning.
4. Return the rotated array.

---

## Code

```python
def rotate_right(arr, k):

    for _ in range(k):

        last = arr.pop()

        arr.insert(0, last)

    return arr


arr = [1,2,3,4,5]

print(rotate_right(arr,2))
```

Output

```python
[4,5,1,2,3]
```

---

# Dry Run

Original

```text
1 2 3 4 5
```

### First Rotation

```text
5 1 2 3 4
```

### Second Rotation

```text
4 5 1 2 3
```

Final Output

```python
[4,5,1,2,3]
```

---

# Time Complexity (Brute Force)

Each rotation shifts almost every element.

One rotation requires

```text
O(n)
```

Rotating **k** times gives

```text
Time Complexity = O(n × k)
```

---

# Space Complexity

No additional array is created.

```text
Space Complexity = O(1)
```

---

# Can It Be Optimized?

Yes.

The brute-force solution becomes slow when **k** is very large.

Instead of rotating the array one step at a time, we can rotate it using the **Reverse Algorithm**.

---

# Optimal Approach — Reverse Algorithm

This is the standard interview solution.

---

## Idea

A right rotation can be completed using **three reverse operations**.

---

## Algorithm

Suppose

```python
arr = [1,2,3,4,5,6,7]

k = 3
```

### Step 1

Reverse the entire array.

```text
1 2 3 4 5 6 7

↓

7 6 5 4 3 2 1
```

---

### Step 2

Reverse the first **k** elements.

```text
7 6 5

↓

5 6 7
```

Array becomes

```text
5 6 7 4 3 2 1
```

---

### Step 3

Reverse the remaining elements.

```text
4 3 2 1

↓

1 2 3 4
```

Final Array

```text
5 6 7 1 2 3 4
```

---

# Why Does the Reverse Algorithm Work?

The first reverse places the last **k** elements at the beginning, but in reverse order.

The second reverse restores the correct order of those first **k** elements.

The third reverse restores the correct order of the remaining elements.

Together, these three reversals produce the required rotation.

---

# Time Complexity (Optimal)

Three reverse operations are performed.

Each reverse requires

```text
O(n)
```

Therefore

```text
3 × O(n)

↓

O(n)
```

---

# Space Complexity

The reversal is performed within the same array.

No additional array is required.

```text
Space Complexity = O(1)
```

---

# Complexity Comparison

| Method               | Time Complexity | Space Complexity |
| -------------------- | --------------- | ---------------- |
| Brute Force Rotation | O(n × k)        | O(1)             |
| Reverse Algorithm    | O(n)            | O(1)             |

---

# Why Is the Reverse Algorithm Better?

The brute-force approach performs unnecessary repeated rotations.

The reverse algorithm completes the entire rotation in linear time using constant extra memory.

Therefore, it is considered the optimal interview solution.

---

# Common Beginner Mistakes

## Mistake 1 — Ignoring Large Values of k

Suppose

```python
arr = [1,2,3,4,5]

k = 12
```

Rotating 12 times is unnecessary.

Instead,

```python
k = k % len(arr)
```

Example

```python
12 % 5 = 2
```

So rotating 12 times is exactly the same as rotating 2 times.

---

## Mistake 2 — Forgetting to Normalize k

Always write

```python
k = k % len(arr)
```

before performing the rotation.

---

## Mistake 3 — Using Brute Force for Large Inputs

For large arrays, the brute-force method becomes inefficient.

The Reverse Algorithm should be preferred in interviews.

---

# Interview Tips

Whenever an interviewer asks:

> **Rotate an array by k positions**

Think in this order:

1. Normalize `k`

```python
k = k % len(arr)
```

2. Use the **Reverse Algorithm**

Avoid rotating the array one step at a time.
