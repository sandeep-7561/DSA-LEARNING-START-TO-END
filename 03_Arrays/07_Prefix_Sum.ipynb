````python
arr = [2,4,6,8,10]

prefix = [0] * len(arr) # [0,0,0,0,0]
prefix[0] = arr[0] # [2,0,0,0,0]

for i in range(1,len(arr)):
    prefix[i] = prefix[i-1] + arr[i]

print(prefix)
````

# Prefix Sum in Data Structures & Algorithms

## Introduction

Prefix Sum is one of the most important optimization techniques in Data Structures and Algorithms.

Instead of calculating the sum of a range repeatedly, Prefix Sum precomputes cumulative sums once and answers future range sum queries efficiently.

It is commonly used in:

* Range Sum Queries
* Competitive Programming
* Dynamic Programming
* Sliding Window
* Matrix Prefix Sum
* Many LeetCode Problems

---

# What is Prefix Sum?

## Definition

> **A Prefix Sum Array stores the cumulative sum of all elements from the beginning of the array up to the current index.**

Each position in the prefix array represents the total sum from index **0** to that index.

---

# Example

Original Array

```python
arr = [2, 4, 6, 8, 10]
```

Prefix Sum Array

```python
prefix = [2, 6, 12, 20, 30]
```

---

# How Prefix Sum Works

Original Array

```text
Index

0    1    2    3    4

2    4    6    8    10
```

Prefix Array

```text
Index

0    1    2    3    4

2    6    12   20   30
```

Explanation

```text
prefix[0] = 2

prefix[1] = 2 + 4 = 6

prefix[2] = 2 + 4 + 6 = 12

prefix[3] = 2 + 4 + 6 + 8 = 20

prefix[4] = 2 + 4 + 6 + 8 + 10 = 30
```

Each element stores the cumulative sum from the beginning of the array.

---

# Why Do We Need Prefix Sum?

Suppose the array is

```python
arr = [2, 4, 6, 8, 10]
```

Find the sum from index **1** to **3**.

Normal Method

```python
4 + 6 + 8 = 18
```

Now imagine receiving hundreds or thousands of similar queries.

Example

```text
0 → 2

2 → 4

1 → 4

0 → 4
```

Running a loop for every query becomes inefficient.

Prefix Sum solves this problem by preprocessing the array once.

---

# Normal Approach

## Algorithm

1. Start from the left index.
2. Traverse until the right index.
3. Add every element.
4. Return the total.

---

## Code

```python
def range_sum(arr, left, right):

    total = 0

    for i in range(left, right + 1):
        total += arr[i]

    return total


arr = [2,4,6,8,10]

print(range_sum(arr,1,3))
```

Output

```python
18
```

---

# Time Complexity (Normal Approach)

Each query traverses the selected range.

```text
Time Complexity = O(n)
```

If there are many queries, the total running time becomes large.

---

# Building the Prefix Sum Array

## Algorithm

1. Create a new array of the same size.
2. Store the first element directly.
3. For every remaining index:

```text
Current Prefix Sum

=

Previous Prefix Sum

+

Current Element
```

---

## Code

```python
arr = [2,4,6,8,10]

prefix = [0] * len(arr)

prefix[0] = arr[0]

for i in range(1, len(arr)):
    prefix[i] = prefix[i - 1] + arr[i]

print(prefix)
```

Output

```python
[2,6,12,20,30]
```

# Range Sum Query Using Prefix Sum

Suppose

```python
arr = [2,4,6,8,10]
```

Prefix Array

```python
prefix = [2,6,12,20,30]
```

Find the sum from index **1** to **3**.

Instead of using a loop,

```python
answer = prefix[3] - prefix[0]
```

```text
20 - 2 = 18
```

Correct Answer

```python
18
```

---

# Prefix Sum Formula

If

```text
left == 0
```

Then

```python
answer = prefix[right]
```

Otherwise

```python
answer = prefix[right] - prefix[left - 1]
```

This formula allows every range sum query to be answered in constant time.

---

# Time Complexity

## Building Prefix Sum

The array is traversed once.

```text
Time Complexity = O(n)
```

---

## One Range Query

Only one subtraction is performed.

```text
Time Complexity = O(1)
```

---

# Space Complexity

An additional array is created.

```text
Space Complexity = O(n)
```

---

# Complexity Comparison

| Method           | Build Time | Query Time | Space  |
| ---------------- | ---------- | ---------- | ------ |
| Normal Traversal | None       | O(n)       | O(1)   |
| Prefix Sum       | O(n)       | O(1)       | O(n) ✅ |

---

# Why Is Prefix Sum Better?

The preprocessing step requires only one traversal.

After preprocessing, every range sum query is answered in constant time.

This makes Prefix Sum ideal for problems involving multiple range queries.

# Common Beginner Mistakes

## Mistake 1 — Forgetting the First Element

Wrong

```python
for i in range(len(arr)):
    prefix[i] = prefix[i-1] + arr[i]
```

Correct

```python
prefix[0] = arr[0]

for i in range(1, len(arr)):
    prefix[i] = prefix[i-1] + arr[i]
```

---

## Mistake 2 — Incorrect Formula

Wrong

```python
prefix[right] - prefix[left]
```

Correct

```python
prefix[right] - prefix[left - 1]
```

---

## Mistake 3 — Forgetting the `left == 0` Case

When

```text
left == 0
```

Simply return

```python
prefix[right]
```

because there is no previous prefix value.
