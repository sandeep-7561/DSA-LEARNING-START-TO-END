````python
'''
Question 1:
Check whether an array is a palindrome.

Input:
[1,2,3,2,1]

Output:
True
'''
arr = [1,2,3,2,1]

def Palindrome(arr):
    left = 0
    right = len(arr) -1
    while left < right:
        if arr[left] != arr[right]:
            return False
        left += 1
        right -= 1
    return True
            
print(Palindrome(arr))
````

# Two Pointer Technique in Data Structures & Algorithms

## Introduction

The **Two Pointer Technique** is one of the most important problem-solving patterns in Data Structures and Algorithms.

Instead of using nested loops, two pointers are used to traverse the array efficiently, often reducing the time complexity from **O(n²)** to **O(n)**.

The Two Pointer Technique is widely used in coding interviews and competitive programming.

It is commonly applied in:

* Reverse Array
* Palindrome Checking
* Two Sum II (Sorted Array)
* Remove Duplicates
* Move Zeroes
* Merge Sorted Arrays
* Container With Most Water
* Trapping Rain Water

---

# What is the Two Pointer Technique?

## Definition

> **The Two Pointer Technique uses two indices (pointers) to traverse an array or string in order to solve problems efficiently while avoiding unnecessary iterations.**

The pointers move according to the problem's requirements.

---

# Why Do We Use Two Pointers?

A brute-force solution often requires nested loops.

Example:

```python
for i in range(n):
    for j in range(n):
        ...
```

Time Complexity

```text
O(n²)
```

Using two pointers, many of these problems can be solved in

```text
O(n)
```

which is significantly faster.

---

# Types of Two Pointer Technique

There are two major patterns.

## 1. Opposite Direction Pointers

One pointer starts from the beginning of the array.

The other pointer starts from the end.

Example

```text
Left

↓

1 2 3 4 5

        ↑

      Right
```

Both pointers move toward each other.

---

### Movement

```python
left += 1
right -= 1
```

---

### Common Applications

* Reverse Array
* Palindrome Checking
* Two Sum II (Sorted Array)
* Container With Most Water

---

# Example — Reverse Array

```python
arr = [1,2,3,4,5]

left = 0
right = len(arr)-1

while left < right:

    arr[left], arr[right] = arr[right], arr[left]

    left += 1
    right -= 1

print(arr)
```

Output

```python
[5,4,3,2,1]
```

---

# Example — Palindrome Check

```python
arr = [1,2,3,2,1]

def is_palindrome(arr):

    left = 0
    right = len(arr)-1

    while left < right:

        if arr[left] != arr[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome(arr))
```

Output

```python
True
```

---

# 2. Same Direction Pointers

Both pointers start from the beginning of the array.

One pointer moves slowly.

The other moves faster.

Example

```text
Slow

↓

1 1 2 2 3 4 4

↓

Fast
```

The fast pointer scans every element.

The slow pointer keeps track of the required position.

---

### Movement

```python
slow += 1
fast += 1
```

---

### Common Applications

* Remove Duplicates
* Move Zeroes
* Merge Sorted Arrays
* Sliding Window (Advanced)

---

# Example — Remove Duplicates

```python
arr = [1,1,2,2,3,4,4]

def remove_duplicates(arr):

    if len(arr) == 0:
        return []

    slow = 0

    for fast in range(1, len(arr)):

        if arr[fast] != arr[slow]:

            slow += 1
            arr[slow] = arr[fast]

    return arr[:slow+1]

print(remove_duplicates(arr))
```

Output

```python
[1,2,3,4]
```

---

# Example — Two Sum II (Sorted Array)

```python
arr = [2,7,11,15]
target = 9

def two_sum(arr, target):

    left = 0
    right = len(arr)-1

    while left < right:

        current_sum = arr[left] + arr[right]

        if current_sum == target:
            return True

        elif current_sum < target:
            left += 1

        else:
            right -= 1

    return False

print(two_sum(arr, target))
```

Output

```python
True
```

---

# Decision Rules for Two Sum II

```text
Current Sum == Target

↓

Answer Found

-------------------------

Current Sum < Target

↓

Increase Left Pointer

-------------------------

Current Sum > Target

↓

Decrease Right Pointer
```

---

# Time Complexity

Most Two Pointer algorithms traverse the array only once.

```text
Time Complexity = O(n)
```

---

# Space Complexity

No additional data structures are required.

```text
Space Complexity = O(1)
```

---

# Opposite vs Same Direction

| Opposite Direction                       | Same Direction                          |
| ---------------------------------------- | --------------------------------------- |
| One pointer starts from the left         | Both pointers start from the left       |
| One pointer starts from the right        | One pointer is slow, the other is fast  |
| Pointers move toward each other          | Both pointers move forward              |
| Used for Reverse, Palindrome, Two Sum II | Used for Remove Duplicates, Move Zeroes |

---

# When Should You Use Two Pointers?

Consider the Two Pointer Technique whenever the problem involves:

* A sorted array
* Pair searching
* Reversing data
* Palindrome checking
* Removing duplicates
* Moving elements
* Optimizing nested loops

---

# Common Beginner Mistakes

## Mistake 1 — Choosing the Wrong Pattern

Not every problem uses the same pointer movement.

Always decide whether the problem requires:

* Opposite Direction
* Same Direction

---

## Mistake 2 — Forgetting to Move Pointers

Example

```python
left += 1
right -= 1
```

or

```python
slow += 1
fast += 1
```

If pointers are not updated correctly, the algorithm may enter an infinite loop.

---

## Mistake 3 — Comparing the Wrong Values

In problems like **Two Sum II**, compare

```python
arr[left] + arr[right]
```

with the target.

Do **not** compare

```python
arr[left]
```

with

```python
arr[right]
```

---

## Mistake 4 — Using Nested Loops Unnecessarily

Many **O(n²)** solutions can be optimized to **O(n)** using the Two Pointer Technique.

Always look for opportunities to replace nested loops.

---
