# Array Traversal in Data Structures & Algorithms

## Introduction

Array traversal is one of the most fundamental operations in Data Structures and Algorithms.

Almost every array problem begins with traversal.

Examples include:

* Finding the Maximum Element
* Finding the Minimum Element
* Linear Search
* Counting Elements
* Prefix Sum
* Kadane's Algorithm
* Sliding Window
* Two Pointer Technique

Without understanding traversal, solving array problems becomes difficult.

---

# What is Array Traversal?

## Definition

> **Array Traversal is the process of visiting every element of an array exactly once in a systematic order.**

The purpose of traversal is to access, process, or manipulate every element stored in the array.

---

# Example

```python
arr = [10, 20, 30, 40, 50]
```

Traversal Order

```text
10
↓
20
↓
30
↓
40
↓
50
```

Each element is visited exactly once.

---

# Why Do We Need Traversal?

Many array problems require examining every element.

For example:

* Find the largest element
* Find the smallest element
* Count even numbers
* Calculate the sum of all elements
* Search for a specific value

If even one element is skipped, the result may become incorrect.

---

# Basic Array Traversal

```python
arr = [10, 20, 30, 40, 50]

for num in arr:
    print(num)
```

Output

```text
10
20
30
40
50
```

---

# Traversal Using Index

Sometimes we need the position (index) of each element.

```python
arr = [10, 20, 30, 40]

for i in range(len(arr)):
    print(i, arr[i])
```

Output

```text
0 10
1 20
2 30
3 40
```

---

# Why Use Index Traversal?

Index traversal is useful when we need to modify elements.

Example

```python
arr = [10, 20, 30]

for i in range(len(arr)):
    arr[i] += 10

print(arr)
```

Output

```text
[20, 30, 40]
```

Without using indices, the original array cannot be modified directly.

---

# Two Common Traversal Methods

## Method 1 — Value Traversal

```python
for num in arr:
    print(num)
```

### Best Used For

* Reading elements
* Finding maximum
* Finding minimum
* Counting values
* Calculating sum

---

## Method 2 — Index Traversal

```python
for i in range(len(arr)):
    print(arr[i])
```

### Best Used For

* Modifying elements
* Accessing indices
* Comparing neighboring elements
* Swapping values
* Solving interview problems

---

# Reverse Traversal

```python
arr = [10, 20, 30, 40, 50]

for i in range(len(arr)-1, -1, -1):
    print(arr[i])
```

Output

```text
50
40
30
20
10
```

---

# Time Complexity

Suppose an array contains **n** elements.

Traversal visits every element exactly once.

Therefore,

```text
Time Complexity = O(n)
```

---

# Space Complexity

Traversal does not create any additional array or data structure.

Only a single loop variable is used.

Therefore,

```text
Space Complexity = O(1)
```

---

# Why Is Traversal Accepted?

Traversal is accepted because:

* Every element is visited exactly once.
* No element is skipped.
* No unnecessary work is performed.
* The algorithm correctly processes the entire array.

---

# Can It Be Optimized?

If the problem requires visiting every element, the optimal time complexity is already

```text
O(n)
```

It is impossible to process all **n** elements without visiting them.

Therefore, traversal cannot be optimized beyond **O(n)** for such problems.

---

# Common Beginner Mistakes

## Mistake 1

```python
for i in range(len(arr)+1):
    print(arr[i])
```

Problem

```text
IndexError
```

Reason

The last valid index is

```text
len(arr) - 1
```

---

## Mistake 2

```python
for i in range(len(arr)):
    print(i)
```

Output

```text
0
1
2
3
```

This prints indices, **not** the array values.

Correct

```python
print(arr[i])
```

---

## Mistake 3

```python
for num in arr:
    num += 10
```

The original array remains unchanged because `num` is only a temporary variable.

To modify the original array, use index traversal.

---

# Interview Tips

Before writing code, ask yourself:

### Do I only need the values?

Use

```python
for num in arr:
```

---

### Do I also need the indices?

Use

```python
for i in range(len(arr)):
```

Choosing the correct traversal method makes the solution simpler and more efficient.
