# Array Insertion in Data Structures & Algorithms

## Introduction

Array insertion is the process of adding a new element to an array at a specific position.

Depending on where the new element is inserted, the time complexity may vary.

There are three common types of insertion:

* Insertion at the End
* Insertion at the Beginning
* Insertion in the Middle

Understanding how insertion works internally is essential for coding interviews and DSA.

---

# What is Array Insertion?

## Definition

> **Array insertion is the process of adding a new element into an array at a specified position while maintaining the order of the existing elements.**

---

# Types of Array Insertion

## 1. Insertion at the End

Example

```python
arr = [10, 20, 30]

arr.append(40)

print(arr)
```

Output

```python
[10, 20, 30, 40]
```

---

## How It Works

Before

```text
10   20   30
```

Insert

```text
40
```

After

```text
10   20   30   40
```

No existing element needs to be moved.

---

## Time Complexity

```text
O(1)
```

**Interview Note**

In Python, `append()` is **Amortized O(1)**.

Python lists allocate extra memory internally. When the allocated memory becomes full, Python creates a larger array and copies all existing elements into it. Although resizing takes **O(n)** time, it happens infrequently.

Therefore, the average cost of `append()` remains **Amortized O(1)**.

---

# 2. Insertion at the Beginning

Example

```python
arr = [20, 30, 40]

arr.insert(0, 10)

print(arr)
```

Output

```python
[10, 20, 30, 40]
```

---

## Internal Working

Before

```text
Index

0    1    2

20   30   40
```

Need to insert

```text
10
```

at index **0**.

The existing elements cannot be overwritten.

Instead, every element is shifted one position to the right.

```text
20 → Index 1

30 → Index 2

40 → Index 3
```

After shifting

```text
Index

0    1    2    3

10   20   30   40
```

---

## Time Complexity

Every existing element may need to be shifted.

Therefore,

```text
Time Complexity = O(n)
```

---

# 3. Insertion in the Middle

Example

```python
arr = [10, 20, 40, 50]

arr.insert(2, 30)

print(arr)
```

Output

```python
[10, 20, 30, 40, 50]
```

---

## Internal Working

Before

```text
Index

0    1    2    3

10   20   40   50
```

Need to insert

```text
30
```

at index **2**.

Python shifts every element from index **2** onward.

```text
50 → Index 4

40 → Index 3
```

Now insert

```text
30
```

Final array

```text
10   20   30   40   50
```

---

## Time Complexity

Elements after the insertion position must be shifted.

Therefore,

```text
Time Complexity = O(n)
```

---

# Why Are Beginning and Middle Insertions O(n)?

Arrays store elements in **contiguous memory locations**.

When inserting a new element into the beginning or middle, the array must create space by shifting existing elements.

Example

Before

```text
10   20   30   40
```

Insert

```text
15
```

After

```text
10   15   20   30   40
```

To make room for **15**, every element after index **0** must move one position to the right.

This shifting operation requires linear time.

---

# Time Complexity Comparison

| Insertion Position | Time Complexity                  | Reason                                   |
| ------------------ | -------------------------------- | ---------------------------------------- |
| End                | **O(1)** *(Amortized in Python)* | Usually no shifting is required          |
| Beginning          | **O(n)**                         | Every element shifts one position        |
| Middle             | **O(n)**                         | Elements after the insertion point shift |

---

# Space Complexity

Insertion does not require creating an additional array.

Only a single new element is added.

```text
Space Complexity = O(1)
```

---

# Why Is This Approach Accepted?

The insertion operation:

* Places the new element at the correct position.
* Preserves the order of the existing elements.
* Produces the correct array after insertion.

---

# Can It Be Optimized?

For arrays,

**No.**

Because arrays occupy contiguous memory, shifting elements is unavoidable for insertions at the beginning or middle.

If an application requires frequent insertions and deletions, a **Linked List** is generally a better choice.

# Common Beginner Mistakes

## Mistake 1 — Confusing Replacement with Insertion

```python
arr = [10, 20, 30]

arr[1] = 99
```

Output

```python
[10, 99, 30]
```

This is **replacement**, not insertion.

The value **20** is overwritten instead of adding a new element.

---

## Mistake 2 — Confusing `append()` with `insert()`

```python
arr.append(50)
```

Always inserts at the end.

---

```python
arr.insert(index, value)
```

Inserts an element at any specified position.

# Interview Tips

Whenever an interviewer asks about insertion, first determine **where** the insertion occurs.

Ask yourself:

* Is it at the beginning?
* Is it in the middle?
* Is it at the end?

The answer determines the time complexity.

# Key Takeaways

* Array insertion adds a new element at a specific position.
* There are three types of insertion: **End**, **Beginning**, and **Middle**.
* Insertion at the end is **Amortized O(1)** in Python.
* Insertion at the beginning and middle is **O(n)** because elements must be shifted.
* Arrays use contiguous memory, making shifting unavoidable.
* Understanding insertion is essential before learning deletion, rotation, and more advanced array algorithms.
