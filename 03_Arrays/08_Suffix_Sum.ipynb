# Suffix Sum in Data Structures & Algorithms

## Introduction

Suffix Sum is an important preprocessing technique used in Data Structures and Algorithms.

It is the reverse counterpart of **Prefix Sum**.

Instead of storing the cumulative sum from the beginning of the array, Suffix Sum stores the cumulative sum from the current index to the end of the array.

Suffix Sum is commonly used in:

* Dynamic Programming
* Partition Problems
* Product Except Self
* Equilibrium Index
* Competitive Programming
* Prefix + Suffix Based Interview Problems

---

# What is Suffix Sum?

## Definition

> **A Suffix Sum Array stores the cumulative sum of all elements from the current index to the end of the array.**

Each position contains the sum of all elements from that index through the last element.

---

# Example

Original Array

```python
arr = [2, 4, 6, 8, 10]
```

Suffix Sum Array

```python
suffix = [30, 28, 24, 18, 10]
```

---

# How Suffix Sum Works

Original Array

```text
Index

0    1    2    3    4

2    4    6    8    10
```

Suffix Array

```text
Index

0    1    2    3    4

30   28   24   18   10
```

Explanation

```text
suffix[4] = 10

suffix[3] = 8 + 10 = 18

suffix[2] = 6 + 8 + 10 = 24

suffix[1] = 4 + 6 + 8 + 10 = 28

suffix[0] = 2 + 4 + 6 + 8 + 10 = 30
```

Each position stores the sum from the current index to the end of the array.

---

# Why Do We Need Suffix Sum?

Suffix Sum allows us to efficiently access the cumulative sum from any index to the end of the array.

Instead of repeatedly calculating the same sums using loops, we preprocess the array once and reuse the computed values.

It is especially useful when solving problems involving:

* Remaining sum
* Right-side calculations
* Prefix + Suffix combinations

---

# Building the Suffix Sum Array

## Algorithm

1. Create a suffix array of the same size.
2. Store the last element directly.
3. Traverse the array from right to left.
4. At each index:

```text
Current Suffix Sum

=

Next Suffix Sum

+

Current Element
```

---

## Code

```python
arr = [2,4,6,8,10]

suffix = [0] * len(arr)

suffix[-1] = arr[-1]

for i in range(len(arr)-2, -1, -1):
    suffix[i] = suffix[i+1] + arr[i]

print(suffix)
```

Output

```python
[30,28,24,18,10]
```

# Prefix Sum vs Suffix Sum

| Prefix Sum                             | Suffix Sum                           |
| -------------------------------------- | ------------------------------------ |
| Builds from left to right              | Builds from right to left            |
| Stores sum from start to current index | Stores sum from current index to end |
| Starts with the first element          | Starts with the last element         |
| Uses the previous prefix value         | Uses the next suffix value           |

---

# Formula

### Prefix Sum

```python
prefix[i] = prefix[i-1] + arr[i]
```

---

### Suffix Sum

```python
suffix[i] = suffix[i+1] + arr[i]
```

---

# Time Complexity

The array is traversed once.

```text
Time Complexity = O(n)
```

---

# Space Complexity

An additional suffix array is created.

```text
Space Complexity = O(n)
```

---

# Why Is This Approach Accepted?

The algorithm processes each element exactly once while storing the cumulative sum from the current index to the end.

Every suffix value is computed correctly without repeated calculations.

---

# Can It Be Optimized?

Building the suffix array requires storing cumulative sums.

Therefore:

* Best Time Complexity = **O(n)**
* Best Space Complexity = **O(n)**

# Common Beginner Mistakes

## Mistake 1 — Traversing in the Wrong Direction

Wrong

```python
for i in range(len(arr)):
```

Correct

```python
for i in range(len(arr)-2, -1, -1):
```

Suffix Sum must always be built from **right to left**.

---

## Mistake 2 — Using the Prefix Formula

Wrong

```python
suffix[i] = suffix[i-1] + arr[i]
```

Correct

```python
suffix[i] = suffix[i+1] + arr[i]
```

---

## Mistake 3 — Forgetting to Initialize the Last Element

Wrong

```python
suffix = [0] * len(arr)
```

Correct

```python
suffix[-1] = arr[-1]
```

The last suffix value is simply the last array element.
