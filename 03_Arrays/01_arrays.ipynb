# What is an Array?

## Definition

> **An array is a collection of elements stored in contiguous memory locations.**

All elements are stored one after another in memory, allowing fast access using an index.

---

# Example

```python
arr = [10, 20, 30, 40, 50]
```

Visualization

```text
Index

 0     1     2     3     4

┌─────┬─────┬─────┬─────┬─────┐
│ 10  │ 20  │ 30  │ 40  │ 50  │
└─────┴─────┴─────┴─────┴─────┘
```

---

# Indexing

Every element in an array has a unique index.

The first element always starts at index **0**.

Example

```python
arr = [10, 20, 30, 40, 50]

print(arr[0])   # 10

print(arr[3])   # 40
```

---

# Why Does Indexing Start at 0?

Suppose the first element starts at memory address **1000**, and each integer occupies **4 bytes**.

| Index | Address |
| ----: | ------: |
|     0 |    1000 |
|     1 |    1004 |
|     2 |    1008 |
|     3 |    1012 |

The address of any element can be calculated using:

```text
Address = Base Address + (Index × Size of Data Type)
```

Because the first element has an offset of **0**, array indexing starts from **0**.

---

# Array Access

Example

```python
arr = [10, 20, 30, 40]

print(arr[2])
```

Output

```text
30
```

---

# Time Complexity of Access

Accessing an element by its index takes constant time.

```text
Time Complexity = O(1)
```

The programming language directly calculates the memory address and retrieves the element without searching.

---

# Why Are Arrays Fast?

Arrays support **direct access**.

Unlike linear search, the computer does not visit previous elements.

It immediately jumps to the required memory location.

This is why array indexing is extremely efficient.

---

# Common Array Operations

| Operation           | Time Complexity  |
| ------------------- | ---------------- |
| Access by Index     | O(1)             |
| Traversal           | O(n)             |
| Linear Search       | O(n)             |
| Insertion at End*   | O(1) (Amortized) |
| Insertion in Middle | O(n)             |
| Deletion            | O(n)             |
| Reverse             | O(n)             |
| Rotation            | O(n)             |

> **Note:** In Python, `list.append()` is **amortized O(1)** because the list occasionally resizes its internal storage.
