# Sliding Window Technique — DSA Notes

## 1. Introduction

The **Sliding Window Technique** is an important optimization pattern used in Data Structures and Algorithms.

It is mainly used when a problem involves a **contiguous subarray or substring**.

Instead of repeatedly calculating every possible subarray from scratch, we maintain a window and move it across the array while updating the required information.

The Sliding Window Technique can often reduce a brute-force solution from **O(n²)** to **O(n)**.

---

# 2. What is a Sliding Window?

A **window** represents a continuous portion of an array or string.

Example:

```python
arr = [2, 4, 6, 8, 10]
```

Suppose:

```python
k = 3
```

The first window is:

```text
[2, 4, 6] 8 10
```

After moving one position:

```text
2 [4, 6, 8] 10
```

After moving again:

```text
2 4 [6, 8, 10]
```

The window is continuously **sliding** from left to right.

---

# 3. Why Do We Use Sliding Window?

Consider:

```python
arr = [2, 4, 6, 8, 10]
k = 3
```

We want the maximum sum of any subarray of size `3`.

Possible windows:

```text
2 + 4 + 6 = 12

4 + 6 + 8 = 18

6 + 8 + 10 = 24
```

Answer:

```text
24
```

A brute-force solution recalculates the entire sum for every window.

This creates unnecessary work.

Sliding Window avoids this repeated calculation.

---

# 4. The Main Idea

Suppose the current window is:

```text
[2, 4, 6]
```

Its sum is:

```text
12
```

The next window is:

```text
[4, 6, 8]
```

Instead of calculating:

```text
4 + 6 + 8
```

again, we simply:

1. Remove the element leaving the window.
2. Add the new element entering the window.

Formula:

```text
New Window Sum
=
Old Window Sum
- Outgoing Element
+ Incoming Element
```

Example:

```text
12 - 2 + 8 = 18
```

This is the core idea behind the Sliding Window technique.

---

# 5. Fixed-Size Sliding Window

In a **Fixed-Size Sliding Window**, the window size remains constant.

Example:

```python
k = 3
```

Every window will contain exactly 3 elements.

```text
[2,4,6] 8 10

2 [4,6,8] 10

2 4 [6,8,10]
```

The window size is always `3`.

---

# 8. Understanding `arr[i-k]`

This is one of the most important parts of the code:

```python
window_sum = window_sum - arr[i-k] + arr[i]
```

Why `i-k`?

Because `arr[i-k]` is the element that is **leaving the window**.

Example:

```text
arr = [2,4,6,8,10]
k = 3
```

When:

```python
i = 3
```

Then:

```python
i - k = 3 - 3 = 0
```

So:

```python
arr[i-k]
```

becomes:

```python
arr[0]
```

which is:

```text
2
```

That is exactly the element leaving the window.

At the same time:

```python
arr[i]
```

is:

```python
arr[3]
```

which is:

```text
8
```

the new element entering the window.

Therefore:

```text
Old Sum - 2 + 8
```

# 9. Brute Force vs Sliding Window

## Brute Force

For every window, calculate all `k` elements again.

```text
Time Complexity = O(n × k)
```

---

## Sliding Window

Calculate the first window once, then update the sum while sliding.

```text
Time Complexity = O(n)
```

---

# 10. Complexity Comparison

| Approach       | Time Complexity | Space Complexity |
| -------------- | --------------: | ---------------: |
| Brute Force    |        O(n × k) |             O(1) |
| Sliding Window |            O(n) |           O(1)   |

---

# 11. Space Complexity

The Sliding Window solution does not create another array.

It only uses variables such as:

```python
window_sum
max_sum
i
```

Therefore:

```text
Space Complexity = O(1)
```

# 12. Fixed-Size Sliding Window Pattern

The general pattern is:

```python
def sliding_window(arr, k):

    window_sum = sum(arr[:k])
    answer = window_sum

    for i in range(k, len(arr)):

        window_sum -= arr[i-k]
        window_sum += arr[i]

        answer = max(answer, window_sum)

    return answer
```

The important pattern is:

```text
Create first window
        ↓
Calculate its result
        ↓
Remove outgoing element
        ↓
Add incoming element
        ↓
Update answer
        ↓
Slide again
```

* Maximum/Minimum Element in Every Window
* Count occurrences within a window

# 14. Variable-Size Sliding Window

A **Variable-Size Sliding Window** is different from a fixed-size window.

Here, the window size is **not fixed**.

Instead, the window expands or shrinks according to a condition.

Example:

```text
Left
 ↓
[ 2  1  5  1  3  2 ]
                   ↑
                  Right
```

The window may become:

```text
[2,1]
```

then:

```text
[2,1,5]
```

and later shrink:

```text
[1,5]
```

The size changes dynamically.

---

# 15. Fixed vs Variable Sliding Window

| Fixed Size           | Variable Size             |
| -------------------- | ------------------------- |
| Window size is known | Window size changes       |
| Usually given `k`    | Usually given a condition |
| Example: size `k`    | Example: sum ≥ target     |
| Expand and slide     | Expand and shrink         |
| Usually simpler      | More advanced             |

---

# 17. Variable-Size Sliding Window Pattern

Think:

```text
Expand Right
        ↓
Condition satisfied?
        ↓
Try shrinking from Left
        ↓
Update Answer
```

General structure:

```python
left = 0

for right in range(len(arr)):

    # Add arr[right]

    while condition_is_invalid():

        # Remove arr[left]
        left += 1

    # Update answer
```

This pattern will be extremely important for advanced LeetCode problems.

# 19. Common Beginner Mistakes

### Mistake 1 — Recalculating the Entire Window

Wrong idea:

```text
Every window → calculate all elements again
```

This can lead to:

```text
O(n × k)
```

Instead:

```text
Old Sum
- Outgoing
+ Incoming
```

---

### Mistake 2 — Forgetting the Outgoing Element

When the window moves, one element leaves.

You must remove it:

```python
window_sum -= arr[i-k]
```

---

### Mistake 3 — Forgetting the Incoming Element

You must also add the new element:

```python
window_sum += arr[i]
```

---

### Mistake 4 — Confusing Sliding Window with Two Pointers

Sliding Window often uses two boundaries:

```text
left
right
```

but its main purpose is to maintain a **contiguous window** while expanding, shrinking, or sliding it.

Two Pointer is a broader technique.

* Sliding Window is an optimization technique for contiguous arrays and strings.
