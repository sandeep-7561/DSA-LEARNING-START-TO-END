````python
#two pointer

arr  = [10,20,30,40,50]

def reverse(arr):
    left = 0
    right = len(arr)-1
    while left < right:
        arr[left],arr[right] = arr[right],arr[left]
        left += 1
        right -= 1
    return arr

print(reverse(arr))
````

# What is Array Reverse?

## Definition

> **Array reversal is the process of rearranging the elements of an array in the opposite order.**

The first element becomes the last, and the last element becomes the first.

---

# Methods to Reverse an Array

There are two common approaches:

1. Reverse Using an Extra Array
2. Reverse In-Place (Two Pointer Technique)

---

# Method 1 — Reverse Using an Extra Array

## Idea

* Create a new empty array.
* Traverse the original array from the last element to the first.
* Append each element into the new array.
* Return the new reversed array.

---

## Algorithm

1. Create an empty array.
2. Start from the last index.
3. Append every element into the new array.
4. Return the new array.

---

## Code

```python
def reverse_array(arr):

    result = []

    for i in range(len(arr) - 1, -1, -1):
        result.append(arr[i])

    return result


arr = [10, 20, 30, 40, 50]

print(reverse_array(arr))
```

Output

```python
[50, 40, 30, 20, 10]
```

# Time Complexity

The loop visits every element exactly once.

```text
Time Complexity = O(n)
```

---

# Space Complexity

A new array is created to store all elements.

```text
Space Complexity = O(n)
```

---

# Can It Be Optimized?

Yes.

Instead of creating a new array, we can reverse the original array itself.

This is called **In-Place Reversal**.

---

# Method 2 — Reverse In-Place (Two Pointer Technique)

## Idea

Use two pointers.

* Left pointer starts from the beginning.
* Right pointer starts from the end.
* Swap both elements.
* Move both pointers toward the center.
* Continue until both pointers meet.

---

## Algorithm

1. Initialize `left = 0`.
2. Initialize `right = len(arr) - 1`.
3. While `left < right`:

   * Swap `arr[left]` and `arr[right]`.
   * Increment `left`.
   * Decrement `right`.
4. Return the reversed array.

---

## Code

```python
def reverse_array(arr):

    left = 0
    right = len(arr) - 1

    while left < right:

        arr[left], arr[right] = arr[right], arr[left]

        left += 1
        right -= 1

    return arr


arr = [10, 20, 30, 40, 50]

print(reverse_array(arr))
```

Output

```python
[50, 40, 30, 20, 10]
```

# Why Do We Use `while left < right`?

When both pointers meet at the middle element, that element is already in its correct position.

Example

```text
1 2 3 4 5
```

The middle element is

```text
3
```

Swapping it with itself is unnecessary.

Therefore,

```python
while left < right
```

is the preferred condition.

---

# Why Is the Two Pointer Method Better?

Method 1

```text
Time  = O(n)

Space = O(n)
```

Method 2

```text
Time  = O(n)

Space = O(1)
```

The second method performs the same work while using constant extra memory.

Therefore, it is the preferred interview solution.

---

# Comparison

| Method      | Time Complexity | Space Complexity |
| ----------- | --------------- | ---------------- |
| Extra Array | O(n)            | O(n)             |
| Two Pointer | O(n)            | O(1)             |

---

# Why Is This Solution Accepted?

The algorithm:

* Swaps the first and last elements.
* Swaps the second and second-last elements.
* Continues until the center is reached.
* Every element reaches its correct reversed position.

---

# Can It Be Optimized?

The time complexity cannot be improved.

Every element must be processed at least once.

```text
Best Time Complexity = O(n)
```

The best possible space complexity is

```text
O(1)
```

using the Two Pointer approach.

# Common Beginner Mistakes

## Mistake 1 — Incorrect Loop Condition

```python
while left <= right:
```

This is unnecessary because the middle element would be swapped with itself.

Preferred solution

```python
while left < right:
```

---

## Mistake 2 — Forgetting to Move Pointers

```python
left += 1
right -= 1
```

Without updating the pointers, the loop never terminates.

---

## Mistake 3 — Incorrect Swapping

Wrong

```python
arr[left] = arr[right]
```

This overwrites data.

Correct

```python
arr[left], arr[right] = arr[right], arr[left]
```

# Key Takeaways

* Array reversal changes the order of elements from first-to-last and last-to-first.

  * Using the Two Pointer Technique.
