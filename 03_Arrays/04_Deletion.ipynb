# Array Deletion in Data Structures & Algorithms

## Introduction

Array deletion is one of the fundamental operations in Data Structures and Algorithms.

Deletion removes an existing element from an array while preserving the order of the remaining elements.

Depending on the position of the deleted element, the time complexity may vary.

---

# What is Array Deletion?

## Definition

> **Array deletion is the process of removing an element from a specified position while maintaining the order of the remaining elements.**

---

# Types of Array Deletion

There are three common types of deletion:

* Deletion from the End
* Deletion from the Beginning
* Deletion from the Middle

---

# 1. Deletion from the End

Example

```python
arr = [10, 20, 30, 40]

arr.pop()

print(arr)
```

Output

```python
[10, 20, 30]
```

---

## How It Works

Before

```text
10   20   30   40
```

Delete

```text
40
```

After

```text
10   20   30
```

No element is shifted.

The last element is simply removed.

---

## Time Complexity

```text
O(1)
```

Reason:

No shifting of elements is required.

---

# 2. Deletion from the Beginning

Example

```python
arr = [10, 20, 30, 40]

arr.pop(0)

print(arr)
```

Output

```python
[20, 30, 40]
```

---

## Internal Working

Before

```text
Index

0    1    2    3

10   20   30   40
```

Delete

```text
10
```

Arrays cannot contain empty gaps.

Therefore, every remaining element shifts one position to the left.

```text
20 → Index 0

30 → Index 1

40 → Index 2
```

Final Array

```text
20   30   40
```

---

## Time Complexity

Every remaining element must be shifted.

```text
Time Complexity = O(n)
```

---

# 3. Deletion from the Middle

Example

```python
arr = [10, 20, 30, 40, 50]

arr.pop(2)

print(arr)
```

Output

```python
[10, 20, 40, 50]
```

---

## Internal Working

Before

```text
Index

0    1    2    3    4

10   20   30   40   50
```

Delete

```text
30
```

The array cannot leave an empty space.

Elements after the deleted position move one position to the left.

```text
40 → Index 2

50 → Index 3
```

Final Array

```text
10   20   40   50
```

---

## Time Complexity

Elements after the deleted position must shift.

```text
Time Complexity = O(n)
```

---

# Why Are Beginning and Middle Deletions O(n)?

Arrays store data in **contiguous memory locations**.

When an element is deleted from the beginning or middle, the remaining elements must shift to remove the empty space.

Example

Before

```text
10   20   30   40   50
```

Delete

```text
20
```

Shift

```text
30 → Left

40 → Left

50 → Left
```

After

```text
10   30   40   50
```

Since shifting depends on the number of remaining elements, the operation takes linear time.

---

# Time Complexity Comparison

| Deletion Position | Time Complexity | Reason                                    |
| ----------------- | --------------- | ----------------------------------------- |
| End               | **O(1)**        | No shifting required                      |
| Beginning         | **O(n)**        | All remaining elements shift              |
| Middle            | **O(n)**        | Elements after the deleted position shift |

---

# Space Complexity

Deletion does not create an additional array or data structure.

Only one element is removed.

```text
Space Complexity = O(1)
```

---

# Why Is This Approach Accepted?

The deletion algorithm:

* Removes the required element.
* Preserves the order of the remaining elements.
* Maintains contiguous memory.
* Produces the correct final array.

---

# Can It Be Optimized?

For arrays,

**No.**

Shifting elements is unavoidable because arrays use contiguous memory.

If an application frequently performs insertions and deletions, a **Linked List** is generally a better choice.

# Common Beginner Mistakes

## Mistake 1 — Confusing `pop()` and `remove()`

### `pop(index)`

Removes an element using its index.

```python
arr = [10, 20, 30]

arr.pop(1)

print(arr)
```

Output

```python
[10, 30]
```

---

### `remove(value)`

Removes an element using its value.

```python
arr = [10, 20, 30]

arr.remove(20)

print(arr)
```

Output

```python
[10, 30]
```

---

## Mistake 2 — Invalid Index

```python
arr.pop(10)
```

If the index does not exist, Python raises

```text
IndexError
```

Always verify that the index is within the valid range.

---

# Interview Tips

Whenever an interviewer asks about deletion, first determine:

* Are we deleting by **index** or by **value**?
* Is the deletion at the **beginning**, **middle**, or **end**?

The answer determines the algorithm and its time complexity.

# Interview Question

Suppose an array contains **1,000,000 elements**.

Which operation is faster?

```python
arr.pop()
```

or

```python
arr.pop(0)
```

### Answer

```python
arr.pop()
```

Reason:

* `pop()` removes the last element without shifting any data → **O(1)**.
* `pop(0)` shifts every remaining element one position to the left → **O(n)**.

---

# Key Takeaways

* Array deletion removes an element while preserving the order of the remaining elements.
* There are three deletion types: **End**, **Beginning**, and **Middle**.
* Deletion from the end is **O(1)**.
* Deletion from the beginning and middle is **O(n)** because elements must shift.
* Arrays cannot contain empty spaces due to contiguous memory allocation.
* Understanding deletion is essential before learning reverse, rotation, sliding window, and advanced array algorithms.
